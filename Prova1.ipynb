Qual estrutura de dados estudada é mais adequada para guradar os chamados já atendidos, considerando que o sistema precisa desfazer o primeiro atendimento mais recente. acredito que a estrutura mais adequada seja pilha, pois no caso da ação "desfazer", o chamado mais recente sai primeiro.
Sem executar o código, indique como estarão as estruturas após a sequência atender atender desfazer atender aguradando na fila 103 e 103 e na pilha 101 e 102

```python
#Parte 2 implementação

class CentralAtendimento:
    def __init__(self):
        self.aguardando = []  # fila: entra no fim, sai do início
        self.atendidos = []   # pilha: entra no topo, sai do topo

    def adicionar_chamado(self, codigo, cliente, prioridade):
        chamado = Chamado(codigo, cliente, prioridade)
        self.aguardando.append(chamado)
        return chamado

    def atender_proximo(self):
        if not self.aguardando:
            print("Não há chamados aguardando.")
            return None
        chamado = self.aguardando.pop(0)
        self.atendidos.append(chamado)
        return chamado

    def desfazer_ultimo_atendimento(self):
        if not self.atendidos:
            print("Não há atendimentos para desfazer.")
            return None
        chamado = self.atendidos.pop()
        self.aguardando.insert(0, chamado)
        return chamado

    def buscar_chamado(self, codigo):
        for chamado in self.aguardando:
            if chamado.codigo == codigo:
                print(f"Chamado {codigo} ({chamado.cliente}): aguardando atendimento")
                return chamado
        for chamado in self.atendidos:
            if chamado.codigo == codigo:
                print(f"Chamado {codigo} ({chamado.cliente}): já atendido")
                return chamado
        print(f"Chamado {codigo} não encontrado.")
        return None

    def exibir_situacao(self):
        print("Aguardando:", [c.codigo for c in self.aguardando])
        print("Atendidos: ", [c.codigo for c in self.atendidos])
```
